```python
!pip install scikeras
!pip install scikit-learn==1.4.2
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from sklearn.model_selection import train_test_split, StratifiedKFold,cross_val_score
from scikeras.wrappers import KerasClassifier
```

```python
data = pd.read_csv('Churn_Modelling.csv')
data.head()
```

```python
data = data.drop (["RowNumber","CustomerId","Surname"],axis = 1)
```

```python
# Convert Categorical Data
data = pd.get_dummies(data,columns = ["Geography","Gender"],drop_first =True)
```

```python
# Taking input and out put from the datset
x = data.drop ("Exited",axis = 1)
y = data ["Exited"]
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x = sc.fit_transform(x)
```

```python
# creating ANN
def create_model ():
  model = Sequential ()
  model.add (Dense (units = 11,activation = "relu", input_dim=x.shape[1]))
  model.add (Dense (units = 6,activation = "relu"))
  model.add (Dense (units = 1,activation = "sigmoid"))
  model.compile (optimizer = Adam(learning_rate = 0.01),loss = "binary_crossentropy",metrics = ["accuracy"])
  return model
```

```python
# Wrap ANN for scikit learn
ann = KerasClassifier (model = create_model, epochs = 20,batch_size = 32)
```

```python
# Create 5 folds
kfold = StratifiedKFold (n_splits = 5, shuffle= True,random_state = 42)
```

```python
# Performs Cross Validation
scores = cross_val_score (ann,x,y,cv = kfold)
```

```python
print ("Fold 1:", scores [0])
print ("Fold 2:", scores [1])
print ("Fold 3:", scores [2])
print ("Fold 4:", scores [3])
print ("Fold 5:", scores [4])
print ("Average Accuracy:",np.mean (scores))
```
